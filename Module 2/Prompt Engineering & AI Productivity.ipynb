```python
# 1
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()

zero_shot_prompt = PromptTemplate(template='You are a helpful assistant {query},answer in 50 words only',
input_variables =['query'])
few_shot_prompt = PromptTemplate(template='You are a helpful assistant {query}, examples {examples},answer in 50 words only',
                                 input_variables=['query','examples'])

LLM = ChatGroq(model='openai/gpt-oss-20b')
def prompt(type='zero_shot_prompt'):
    if type=='zero_shot_prompt':
        chain = zero_shot_prompt | LLM | StrOutputParser()
    else:
        chain = few_shot_prompt | LLM | StrOutputParser()
    return chain
res1 = prompt('zero_shot_prompt').invoke({'query':'Tell me how to call me a taxi'})
res2 = prompt('few_shot_prompt').invoke({'query':'Tell me how to study hard','examples':'1. To study hard make motivation 2. To study hard make sure to be consistant'})
print(res1,'\n\n',res2)
```

```python
from pydantic import BaseModel,Field
class output(BaseModel):
    Material_No: int = Field(description='This consist of number of total materials needed for the work'),
    Material_UseCase: str = Field(description='Consist of short summary of the materials required to needed to make'),
    final_product: str = Field(description='Name of the final product that will be made in Maximum 5 words')
from langchain_core.output_parsers import PydanticOutputParser
parser = PydanticOutputParser(pydantic_object=output)
prompt = PromptTemplate(template='You are a helpful assistant help me with this {query},  Follow these formatting instructions: {format_instructions}',
                        input_variables=['query'],
                        partial_variables={'format_instructions':parser.get_format_instructions()})
chain = prompt | LLM | parser

res3 = chain.invoke({'query':'How to make a Tesla Car'})
print(res3)
    
```

```python
prompt = PromptTemplate(
    template="""
    You are an expert Python developer.
    Task: {task}
    Input: {query}
    Provide a simple and clear solution.
    """,
    input_variables=["task", "query"]
)

chain = prompt | LLM | StrOutputParser()

# Test cases
test_cases = [
    {"task": "Generate Code", "query": "Write Python code to reverse a string"},
    {"task": "Debug Code", "query": "Fix this code: print(10/0)"},
    {"task": "Explain Code", "query": "Explain Python for loop"}
]

# Evaluate prompts and generate report
for i, test in enumerate(test_cases, 1):
    result = chain.invoke(test)

    print(f"\nTest Case {i}")
    print("Task:", test["task"])
    print("Result:", result)
    print("-" * 50)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Reusable AI function
def ai_tool(task, query):
    prompt = PromptTemplate(
        template="""
        You are a helpful AI assistant.

        Task: {task}
        Input: {query}

        Provide a useful and clear response.
        """,
        input_variables=["task", "query"]
    )

    chain = prompt | LLM | StrOutputParser()

    return chain.invoke({
        "task": task,
        "query": query
    })


# 1. Summarization
summary = ai_tool(
    "Summarize",
    "Artificial Intelligence is transforming the world..."
)
print("Summary:", summary)


# 2. Email Writing
email = ai_tool(
    "Write Email",
    "Write an email requesting leave for two days."
)
print("Email:", email)


# 3. Resume Improvement
resume = ai_tool(
    "Improve Resume",
    "I know Python, Java, LangChain and RAG."
)
print("Resume:", resume)


# 4. Coding Assistance
code = ai_tool(
    "Generate Code",
    "Write Python code to find factorial of a number."
)
print("Code:", code)


# 5. Research Assistance
research = ai_tool(
    "Research",
    "Explain the applications of Generative AI."
)
print("Research:", research)
```
